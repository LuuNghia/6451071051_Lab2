#Bài 2 (Trang 32): Tính toán giá trị F-measure của các bài tập cơ bản

**Công thức F-measure:**
$$\text{F-measure} = 2 \times \frac{\text{precision} \times \text{recall}}{\text{precision} + \text{recall}}$$

#1 Định nghĩa hàm, nạp số liệu và tính toán bảng tổng hợp

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# 1. Hàm tính F-measure theo công thức trung bình điều hòa
def calculate_f_measure(precision, recall):
    return 0.0 if (precision + recall) == 0 else round(2 * (precision * recall) / (precision + recall), 4)

# 2. Bảng dữ liệu Precision, Recall và tính trực tiếp F-measure
data_summary = [
    # Bài 2 - Adult Income
    {"Bài tập": "Bài 2 - Adult Income", "Mô hình": "Decision Tree (ID3)", "Precision": 0.88, "Recall": 0.88},
    {"Bài tập": "Bài 2 - Adult Income", "Mô hình": "Decision Tree (CART)", "Precision": 0.88, "Recall": 0.88},
    {"Bài tập": "Bài 2 - Adult Income", "Mô hình": "Naïve Bayes", "Precision": 0.82, "Recall": 0.93},

    # Bài 4 - Red Wine Quality
    {"Bài tập": "Bài 4 - Red Wine Quality", "Mô hình": "Decision Tree (ID3)", "Precision": 0.61, "Recall": 0.61},
    {"Bài tập": "Bài 4 - Red Wine Quality", "Mô hình": "Decision Tree (CART)", "Precision": 0.62, "Recall": 0.62},
    {"Bài tập": "Bài 4 - Red Wine Quality", "Mô hình": "Naïve Bayes", "Precision": 0.54, "Recall": 0.53},

    # Bài 5 - Heart Disease
    {"Bài tập": "Bài 5 - Heart Disease", "Mô hình": "Decision Tree (ID3)", "Precision": 0.71, "Recall": 0.76},
    {"Bài tập": "Bài 5 - Heart Disease", "Mô hình": "Decision Tree (CART)", "Precision": 0.75, "Recall": 0.78},
    {"Bài tập": "Bài 5 - Heart Disease", "Mô hình": "Naïve Bayes", "Precision": 0.81, "Recall": 0.85}
]

df_summary = pd.DataFrame(data_summary)
df_summary["F-measure"] = df_summary.apply(lambda r: calculate_f_measure(r["Precision"], r["Recall"]), axis=1)

print("--- BẢNG GIÁ TRỊ F-MEASURE CÁC BÀI TẬP CƠ BẢN ---")
display(df_summary)
```

#2 Trực quan hóa đồ thị Barplot

```python
# In thông tin sinh viên
print("Họ và tên: Lưu Đình Nghĩa")
print("Mã sinh viên: 6451071051")
print("-" * 30)

# Vẽ biểu đồ cột so sánh
plt.figure(figsize=(12, 6))
sns.barplot(data=df_summary, x="Bài tập", y="F-measure", hue="Mô hình", palette="Set2")
plt.title("So sánh giá trị F-measure giữa các mô hình trong các bài tập cơ bản", fontsize=14, fontweight="bold")
plt.ylabel("Giá trị F-measure (F1-score)", fontsize=11)
plt.xlabel("Bài tập thực nghiệm", fontsize=11)
plt.ylim(0, 1.15)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend(title="Mô hình", loc="lower right")
plt.tight_layout()
plt.show()

```
